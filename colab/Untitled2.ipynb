```python
df = pd.DataFrame({"yyyy-mm-dd" : ["2005-09-28", "2007-10-05", "2013-12-20"]})
df
```

```python
data = "2005-12-20"

data[0:4]

data.split("-")
data.split("-")[0]
```

```python
#  년도 컬럼 추가

# 새로운 컬럽 추가
# df['year'] = [2005, 2007, 2012]

def extract_year(column):
  return data.split("-")[0]
```

```python
# df에 extract_year() 함수 적용
# lamda 함수도 가능

df['year'] = df['yyyy-mm-dd'].apply(extract_year)
df
```
